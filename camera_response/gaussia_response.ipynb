```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
center = 8000
sigma = 10

x = np.linspace(center - 3 * sigma, center + 3 * sigma, 21)
y = np.exp(-1 * ((x - center) / sigma) ** 2) / (sigma * np.sqrt(2 * np.pi))
y = (y-y.min()) / (y.max() - y.min()) 

plt.figure(figsize=(10, 6))
plt.plot(x, y)
plt.title("Gaussian Distribution")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid()
```
